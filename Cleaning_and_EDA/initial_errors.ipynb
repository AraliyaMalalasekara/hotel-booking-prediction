```python
import pandas as pd

df_1 = pd.read_csv('drive/MyDrive/Data_Science/Hotel-A-test.csv')
df_2 = pd.read_csv('drive/MyDrive/Data_Science/Hotel-A-train.csv')
df_3 = pd.read_csv('drive/MyDrive/Data_Science/Hotel-A-validation.csv')
```

```python
print("Test shape:", df_1.shape)
print("Train shape:", df_2.shape)
print("Validation shape:", df_3.shape)
```

```python
#Checking missing values
print("Missing values in Test:")
print(df_1.isnull().sum())

print("Missing values in Train:")
print(df_2.isnull().sum())

print("Missing values in Validation:")
print(df_3.isnull().sum())
```

```python
#Checking the columns across the datasets
print("Test columns:", df_1.columns)
print("Train columns:", df_2.columns)
print("Validation columns:", df_3.columns)
```

Since the test dataset doesn't has a column as "Reservation_status", that could be a problem for the model that we are creating.

```python
print(df_1["Booking_date"].str.len().value_counts())
print(df_2["Booking_date"].str.len().value_counts())
print(df_3["Booking_date"].str.len().value_counts())
```

```python
#Feature mismatch
print("Columns in test but not in train:")
print(set(df_1.columns) - set(df_2.columns))

print("Columns in train but not in validation:")
print(set(df_2.columns) - set(df_3.columns))

print("Columns in validation but not in train:")
print(set(df_3.columns) - set(df_1.columns))
```

```python
#Checking for duplicates

print("Duplicates in Test:", df_1.duplicated().sum())
print("Duplicates in Train:", df_2.duplicated().sum())
print("Duplicates in Validation:", df_3.duplicated().sum())
```

```python
#Checking for datatypes

print("Test datatypes:",df_1.dtypes)
print("Train datatypes:", df_2.dtypes)
print("Validation datatypes:", df_3.dtypes)
```

```python
#Checking for value consistency in categorical columns

print(df_1["Gender"].unique())
print(df_2["Gender"].unique())
print(df_3["Gender"].unique())
```
