```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
train=pd.read_csv('/content/drive/MyDrive/Data Science project lifecycle/cleaned files/Hotel_A_Train_Cleaned.csv')
```

```python
test= pd.read_csv('/content/drive/MyDrive/Data Science project lifecycle/cleaned files/Hotel_A_Test_Cleaned.csv')
```

```python
val=pd.read_csv('/content/drive/MyDrive/Data Science project lifecycle/cleaned files/Hotel_A_Validation_Cleaned.csv')
```

```python
cols_to_drop = ['Reservation-id', 'Expected_checkin', 'Expected_checkout', 'Booking_date']
train = train.drop(columns=cols_to_drop, errors='ignore')
```

##Data Splitting

```python
x=train.drop('Reservation_Status',axis=1)
y=train['Reservation_Status']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# --- MODEL 1: Logistic Regression (Your Uploaded Method) ---
log_model = LogisticRegression(max_iter=1000, random_state=42)
log_model.fit(x_train_scaled, y_train)
log_preds = log_model.predict(x_test_scaled)
```

```python
print("--- LOGISTIC REGRESSION RESULTS ---")
print(f"Accuracy: {accuracy_score(y_test, log_preds):.4f}")
print(classification_report(y_test, log_preds))
```

```python
# --- MODEL 2: Random Forest (For Maximum Accuracy) ---
# Random Forests handle complex hotel data much better than linear models
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(x_train, y_train) # Tree models don't strictly need scaled data!
rf_preds = rf_model.predict(x_test)
```

```python
print("\n--- RANDOM FOREST RESULTS ---")
print(f"Accuracy: {accuracy_score(y_test, rf_preds):.4f}")
print(classification_report(y_test, rf_preds))
```

```python
# --- MODEL 2: Random Forest (Addressing Class Imbalance) ---

# Adding class_weight='balanced' forces the model to care about the minority classes!
rf_model_balanced = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced', # <-- THE MAGIC FIX
    random_state=42
)

rf_model_balanced.fit(x_train, y_train)
rf_preds_balanced = rf_model_balanced.predict(x_test)

print("\n--- BALANCED RANDOM FOREST RESULTS ---")
print(f"Accuracy: {accuracy_score(y_test, rf_preds_balanced):.4f}")
print(classification_report(y_test, rf_preds_balanced))
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

print("Applying SMOTE to balance the training data...")
```

```python
# 1. Apply SMOTE ONLY to the training data (never the test data!)
smote = SMOTE(random_state=42)
x_train_smote, y_train_smote = smote.fit_resample(x_train, y_train)

print(f"Old training target counts:\n{y_train.value_counts()}")
print(f"New SMOTE training counts:\n{y_train_smote.value_counts()}")
```

```python
# 2. Set up the "Tuning Grid" (We keep it small so it doesn't take hours to run)
param_grid = {
    'n_estimators': [100, 200],         # Number of trees
    'max_depth': [10, 15, 20],          # How deep the trees can go
    'min_samples_split': [2, 5, 10],    # Prevents trees from memorizing single rows
    'class_weight': ['balanced']        # Keep our balanced weight!
}
```

```python
# 3. Initialize the Grid Search
print("\nStarting Hyperparameter Tuning... (This might take a minute or two)")
rf_base = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(
    estimator=rf_base,
    param_grid=param_grid,
    cv=3,                 # 3-fold cross validation
    scoring='f1_macro',   # Force it to maximize the F1 score for ALL classes, not just accuracy
    n_jobs=-1             # Use all your computer's processors to make it fast
)
```

```python
# Train the tuned model on the SMOTE data
grid_search.fit(x_train_smote, y_train_smote)

# 4. Predict and Evaluate
best_rf = grid_search.best_estimator_
best_preds = best_rf.predict(x_test)

print(f"\n✅ Tuning Complete! Best Parameters chosen: {grid_search.best_params_}")
print("\n--- TUNED & SMOTED RANDOM FOREST RESULTS ---")
print(f"Overall Accuracy: {accuracy_score(y_test, best_preds):.4f}")
print(classification_report(y_test, best_preds))
```
